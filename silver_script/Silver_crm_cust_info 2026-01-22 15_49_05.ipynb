INIT

```python
import pyspark.sql.functions as F
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
from pyspark.sql.functions import lit,trim,col
```

Reading from Bronze

```python
crm_cust_info = spark.read.table("engineering.bronze.crm_cust_info")
display(crm_cust_info.limit(10))
```

```python
for field in crm_cust_info.schema.fields:
    if isinstance(field.dataType, StringType):
        crm_cust_info = crm_cust_info.withColumn(field.name, trim(col(field.name)))
```

NORMALIZATION

```python
crm_cust_info = crm_cust_info.withColumn(
    'cst_marital_status',
    F.when(col('cst_marital_status') == 'M', 'Married')
     .when(col('cst_marital_status') == 'S', 'Single')
     .when(col('cst_marital_status') == 'D', 'Divorced')
     .when(col('cst_marital_status') == 'W', 'Widowed')
     .otherwise(col('cst_marital_status'))
).withColumn(
    'cst_gndr',
    F.when(col('cst_gndr') == 'M', 'Male')
     .when(col('cst_gndr') == 'F', 'Female')
     .otherwise(col('cst_gndr'))
)
display(crm_cust_info)
```
